### Boxplot 
A boxplot is a graph used to show the distribution of numerical data and identify the spread of data and outliers.

In simple words:
👉 Boxplot hume batata hai ki data kis range mein hai, data ka middle part kya hai, aur koi unusual value (outlier) hai ya nahi.

🔹 Main parts of a Boxplot

A boxplot generally shows 5 important values:

1. Minimum → Data ki lowest value
2. Q1 (25th percentile) → Lower quarter of the data
3. Median (Q2) → Data ki middle value
4. Q3 (75th percentile) → Upper quarter of the data
5. Maximum → Data ki highest value

And jo points boxplot se bahar hote hain, unhe usually outliers kehte hain.

Simple example

Suppose students ke marks hain:

10, 20, 25, 30, 35, 40, 45, 100

Yahan 100 baaki marks se kaafi different hai, so boxplot ise outlier ke roop mein show kar sakta hai.


### Boxplot

A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

### Regression Plots

- regplot
- lmplot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
tips
```

```python
# axes level
# In regration plot hue('sx) parameter are not supporytable.
sns.regplot(data=tips,x='total_bill',y='tip')
```

### Multiplot 
A second way to plot Facet plot-> FacetGrid


```python
# figure level - > repplot -> displot-> catplot -> lmplot '
sns.catplot(data=tips, x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
g = sns.FacetGrid(data=tips, col='day', row='time' , hue='smoker')
g.map(sns.boxplot ,'sex','total_bill')
g.add_legend()
```
