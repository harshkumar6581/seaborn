```python
import seaborn as sns 
import matplotlib 
import plotly.express as px
```

```python
tips=sns.load_dataset('tips')
tips
```

```python
# scatter plot -> axes level function 
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex',style='time',size='size')
```

```python
# relplot -> figure level -> square shape 
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
## scatter using replot -> size and hue
```

```python
# style semantics
```

```python
# line plot 
gap = px.data.gapminder()
temp_df=gap[gap['country']=='India']
temp_df
```

```python
# axes level function 
sns.lineplot(data=temp_df,x='year',y='lifeExp')
```

```python
## using relplot 
sns.relplot(data=temp_df,x='year',y='lifeExp',kind='line')
```

```python
## hue -> style 
hrx_df=gap[gap['country'].isin(['India','Brazil','Germany'])]
hrx_df
```

```python
sns.relplot(kind='line',data=hrx_df,x='year',y='lifeExp',hue='country',style='continent',size='continent')
```

```python
sns.lineplot(data=hrx_df,x='year',y='lifeExp',hue='country')
```

```python
## fact plot is work onlt with figure lebel function andd work with rel plot 
# it will not work with scatterplot and line plot 
sns.relplot(data=tips, x='total_bill', y='tip', kind='line', col='sex', row='day')
```

```python
# col wrap 
sns.relplot(kind='line',data=hrx_df,x='year',y='lifeExp',hue='country')
```

```python
gap
sns.relplot(gap,x='lifeExp',y='gdpPercap',kind='scatter')
```

```python
gap
sns.relplot(gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=4)
```

### 2. Distribution Plots 
- used for unvariate analysis 
- used to find out the distribution 
- range of the obserbation 
- central tendency 
- is the data bimodel 
- Are there outliers 

### Plot under distribution plot 
- histplot 
- kdeplot 
- rugplot 


```python
tips
```

```python
## figure level -> displot 
# axes level -> histplot -> kdeplot -> rugplot 
```

```python
## ploting univariate histogram 
tips 
sns.histplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips, x ='total_bill',kind='hist' )
```

```python
## bin parameter 
sns.displot(data=tips, x ='total_bill',kind='hist',bins=19)
```

```python
## its also a possible to visualize the distribution of a categorical variable using the logic of histogram 
## Discrete bins are automatically set for categorical variable 

# countplot 

sns.displot(data=tips, x ='day',kind='hist')
```

```python
## hue parameter 
sns.displot(data=tips, x ='tip',kind='hist',hue='sex')
```

```python
## element -> step
sns.displot(data=tips, x ='tip',kind='hist',hue='sex',element='step')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x='age',kind='hist',element='step',hue='sex')
```

```python
# faceting using col and row not working histplot function 
sns.displot(data=tips ,x='tip',kind='hist',col='sex',element='step')
```

```python
## kdeplot (kernal density estimation )
# Rather than using discrete bins , a KDE plot smooths the obserbation with a gaussian kernal, producing a continous density estimate 
tips 
sns.kdeplot(data=tips,x='total_bill')
```

```python
## hue -> fill
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True)
```

```python
## rugplot 

# plot marginal distribution by drawing ticks along the x and y axes 

## this function is intended to compliment other plots by showing the location of individual obserbation in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

```python
# Bivariate histogram
# A bivariate histogram bins the data within rectangles that tile the plot 
# and then shows the count of observations within each rectangle with the fill color

sns.histplot(data=tips, x='total_bill', y='tip')
```

```python
# Bivariate Kdeplot
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.kdeplot(data=tips, x='total_bill', y='tip')
```
